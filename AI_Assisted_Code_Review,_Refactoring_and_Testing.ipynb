```python
marks = []

for i in range(3):
    marks.append(float(input("Enter mark: ")))

total = 0

for mark in marks:
    total = total + mark

avg = total / 3

if avg >= 90:
    grade = "A"
elif avg >= 80:
    grade = "B"
elif avg >= 70:
    grade = "C"
elif avg >= 60:
    grade = "D"
else:
    grade = "F"

print("Total:", total)
print("Average:", avg)
print("Grade:", grade)
```
